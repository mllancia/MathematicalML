relaxation method

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x): return 1-np.exp(-2*x)
```

```python
x = 1 # init guess
for i in range(10):
    x1 = f(x)
    print(x1)
    if (abs(x-x1)) < 1e-6:
        print("converged at iration: " + str(i+1))
        break
    x=x1
else:
    print("did not converge")
```

```python
x = np.linspace(-1, 2, 100)
plt.plot(x,f(x))
plt.plot(x, x, "r")
plt.grid()
plt.show()
```

Try to converge to 0

```python
x = -0.1 # init guess
for i in range(30):
    x1 = f(x)
    print(x1)
    if (abs(x-x1)) < 1e-6:
        print("converged at iration: " + str(i+1))
        break
    x=x1
else:
    print("did not converge")
```

That didn't work. Because absolute value of f prime at 0 has an derivate of greater than 1, apparently based on tayler series

lets try rearanging f(x)

```python
def f(x): return (-0.5 * np.log(1-x))
```
