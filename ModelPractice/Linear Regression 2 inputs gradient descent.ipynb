```python
import numpy as np
```

```python
def dB_0(beta, x, y):
    return 2 * (beta[0] + beta[1] * x[0] + beta[2] * x[1] - y)
```

```python
def dB_1(beta, x, y):
    return 2 * x[0] * (beta[0] + beta[1] * x[0] + beta[2] * x[1] - y)
```

```python
def dB_2(beta, x, y):
    return 2 * x[1] * (beta[0] + beta[1] * x[0] + beta[2] * x[1] - y)
```

First since I'm not sure how to do this analytically let's solve this with gradient descent

```python
x1 = np.array([-1,1,-2,0])
x2 = np.array([0,3,2,4])
X = np.column_stack((x1,x2))
y = np.array([3, 2, 0, -1])
n = len(x1)
```

Best guess for beta and set parameters

```python
beta = np.array([0.0,0.0,1.0])
maxiter = 1
acceptance = 0.01
learning_rate = 0.25
converged = False
lam = 0
```

```python
for i in range(maxiter):
    dB_0_grad = 0
    dB_1_grad = 0
    dB_2_grad = 0
    for j in range(len(x1)):
        dB_0_grad += dB_0(beta, X[j], y[j])
        dB_1_grad += dB_1(beta, X[j], y[j])
        dB_2_grad += dB_2(beta, X[j], y[j])
    dB_0_grad /= n
    dB_1_grad /= n
    dB_2_grad /= n

    if np.sqrt(dB_0_grad ** 2 + dB_1_grad ** 2 + dB_2_grad ** 2) < acceptance:
        print(f"Converged at iteration {i}")
        converged = True
        break
    else:
        beta = beta - learning_rate * np.array([dB_0_grad, dB_1_grad, dB_2_grad])
```

```python
def dB_0(beta, x, y):
    return  2* (beta[0] + beta[1] * x[0] + beta[2] * x[1] + beta[3] * x[1] ** 2 - y)
```

```python
def dB_1(beta, x, y):
    return 2 * x[0] * (beta[0] + beta[1] * x[0] + beta[2] * x[1] + beta[3] * x[1] ** 2 - y)
```

```python
def dB_2(beta, x, y):
    return 2 * x[1] * (beta[0] + beta[1] * x[0] + beta[2] * x[1] + beta[3] * x[1] ** 2 - y)
```

```python
def dB_3(beta, x, y):
    return 2 * x[1] ** 2 * (beta[0] + beta[1] * x[0] + beta[2] * x[1] + beta[3] * x[1] ** 2 - y)
```

```python
x1 = np.array([2,1,6,3])
x2 = np.array([-4,1,2,-1])
X = np.column_stack((x1,x2))
y = np.array([2,1,5,-2])
n = len(x1)
```

```python
beta = np.array([0.0,0.0,1.0, 1.0])
maxiter = 30000
acceptance = 0.01
learning_rate = 0.001
converged = False
lam = 0
```

```python
for i in range(maxiter):
    dB_0_grad = 0
    dB_1_grad = 0
    dB_2_grad = 0
    dB_3_grad = 0
    for j in range(len(x1)):
        dB_0_grad += dB_0(beta, X[j], y[j])
        dB_1_grad += dB_1(beta, X[j], y[j])
        dB_2_grad += dB_2(beta, X[j], y[j])
        dB_3_grad += dB_3(beta, X[j], y[j])
    dB_0_grad /= n
    dB_1_grad /= n
    dB_2_grad /= n
    dB_3_grad /= n

    if np.sqrt(dB_0_grad ** 2 + dB_1_grad ** 2 + dB_2_grad ** 2 + dB_3_grad ** 2) < acceptance:
        print(f"Converged at iteration {i}")
        converged = True
        break
    else:
        beta = beta - learning_rate * np.array([dB_0_grad, dB_1_grad, dB_2_grad, dB_3_grad])
```
