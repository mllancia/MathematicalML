```python
import numpy as np
```

start with a data matrix

```python
group = np.array([1,1,0,0,0])
x1 = np.array([5,4,6,6,7])
x2 = np.array([2.3, 2.4, 3.0,2.4,2.6])
```

One method of LDA is all about covariance matrices

That is, we need the covariance matrix of every group in isolation

```python
g0_ind = (group == 0)
g1_ind = (group == 1)

x1_g0 = x1[g0_ind]
x2_g0 = x2[g0_ind]
x1_g1 = x1[g1_ind]
x2_g1 = x2[g1_ind]
```

```python
cov_g0 = np.cov(x1_g0, x2_g0)
cov_g1 = np.cov(x1_g1, x2_g1)
```

We sum and average those covariance matrices together to get W the "pooled within-group cov mat"

This can be just the cov groups added together and divided by n the number of groups

or it can be weighted cov matrices by the number of instances within the group - 1 divided by (total instances - number of groups)

```python
W = ( (len(g0_ind) - 1 ) * cov_g0 + (len(g1_ind) - 1) * cov_g1 ) / (len(group) - 2)
```

Plus the cov mat of all the data

```python
T = np.cov(x1, x2)
```

And then we subtract T - W to get B

That's the total cov - the averaged or weighted pooled within group cov mat, which gives us

B the between class scatter matrix, or the between class cov matrix

```python
B = T - W
```

We are looking for S = W^-1 * B

```python
W_inv = np.linalg.inv(W)
```

```python
S = W_inv @ B
```

```python
S
```

We want to maximize this Seperation matrix

Now we do like PCA but instead of on the basic cov matrix of all the data, we do it on S

that means eigenvectors and eigenvalues

```python
eigenvalues, eigenvectors = np.linalg.eig(S)
```

```python
eigenvectors
```

find the greatest eigenvalue: that's your best eigenvector LD1, but make sure to normalize it to length 1

```python
index = np.argmax(eigenvalues)
LD1 = eigenvectors[:, index]
LD1 = LD1 / np.linalg.norm(LD1)
```

```python
X = np.column_stack([x1, x2])
```

```python
scores = X @ LD1
```

```python
score = x1 * LD1[0] + x2 * LD1[1]
```
