```python
import numpy as np
```

```python
def knn (x, y, x_test, y_test, k):
    # get the distances between x_test and xs
    distances = np.abs(x - x_test)
    # use this to find the k nearest neighbours
    nearest_indices = np.argsort(distances)[:k]
    # average them
    knn_avg = np.mean(y[nearest_indices])
    # take the squared error
    return knn_avg
```

```python
def leave_one_out_cross_validation (x, y, k):
    n = len(x)
    total_error = 0
    for i in range(n):
        x_test = x[i]
        y_test = y[i]
        x_train = np.delete(x,i)
        y_train = np.delete(y,i)
        y_pred = knn(x_train, y_train,x_test, y_test, k)
        total_error += (y_test - y_pred) ** 2
    total_error /= n
    return total_error
```

```python
x = np.array([1,3,4,-4,-3,-1])
y = np.array([1,2,5,-3,-4,-1])
```

```python
err = leave_one_out_cross_validation(x,y,2)
```
