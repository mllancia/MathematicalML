```python
import numpy as np
```

```python
def pca_opt (X, d, n_iterations = 1000, lr = .01):
    # d is number of principal components
    N, D = X.shape # rows, columns, or number samples, components

    V_d = np.random.randn(D, d) # initialize a random matrix V of shape D, d

    # Orthonormalize initial V_d (Gram-Schmidt)
    V_d, _ = np.linalg.qr(V_d)

    for iteration in range(n_iterations):
        reconstruction = X @ V_d @ V_d.T @ X.T  # (N, D)
        error = X.T - reconstruction.T

        # Gradient of reconstruction error w.r.t. V_d
        # d/dV_d of ||X - X V_d V_d^T||^2
        gradient = -2 * error.T @ X @ V_d

        V_d -= lr * gradient

        # Re-orthonormalize to maintain constraint
        V_d, _ = np.linalg.qr(V_d)

        if iteration % 100 == 0:
            loss = np.sum(np.linalg.norm(X - X @ V_d @ V_d.T, axis=1)**2)
            print(f"Iteration {iteration}, Loss: {loss:.4f}")

    return V_d
```

```python
x1 = [4, 8, 13, 7]
x2 = [11, 4, 5, 14]
X = np.column_stack((x1,x2))
```

```python
x1_mean = np.mean(x1)
x2_mean = np.mean(x2)
x1_std = np.std(x1)
x2_std = np.std(x2)

x1_norm = ( x1-x1_mean ) / x1_std
x2_norm = ( x2-x2_mean ) / x2_std

X_norm = np.column_stack((x1_norm,x2_norm))
```

```python
pca_opt(X_norm, 1)
```
