start by defining function we want to minimize by optimizing x

We assume that we can't just set the gradient equal to zero and solve for x

We simply take the gradient at the point and go in the direction that gradient says to move at the learning rate

For instance, if gradient is 1, it is moving up, so we need to decrease x learning_rate * (-grad)

```python
def func(x):
    return x**2 + 2*x + 3
```

```python
def grad_func(x):
    return 2*x + 2
```

gradient descent invol

```python
learning_rate = 0.1
x = 1
max_iter = 1000
acceptance = 0.1
converged = False
```

```python
for i in range(max_iter):
    grad = grad_func(x)
    if grad < acceptance:
        converged = True
        print(f"Converged at iteration {i}")
        break
    else:
        x = x - grad * learning_rate
```

```python
print(converged)
print(x)
```
