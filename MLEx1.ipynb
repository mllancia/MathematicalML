```python
# reduct Matrix to L2 Factorization
shape = M.shape
rows = shape[0]
columns = shape[1]

M_fact = M.copy()  # this will be our factorized matrix
b_fact = b.copy()  # this will be our modified vector

for row in range(rows):
    # first, if you have non zero values to the left of the diagonal
    # you need to remove them

    # check every value in the row up until the diagonal
    for column in range(row):
        current_value = M_fact[row, column]
        if current_value != 0:
            # ding ding, we have a winner
            # let's fix it
            # we need to grab the row who's diagonal is directly above,
            # find a constant to multiply our row by, and subtract
            # the row above, multiplied by the constant, from our row
            # such that we get 0 here
            # let's change the name of our variable to make it clear
            # that we're looking at the upper row
            upper_ind = column

            factor = current_value / M_fact[upper_ind, upper_ind]

            # now let's multiple the entire upper row by the constant
            # and subtract it from our row
            for column in range(columns):
                M_fact[row, column] -= factor * M_fact[upper_ind, column]
            # congrats! We'ved (most of) our row
            # let's not forget about the right side of our equation
            # like I did on the BLOCK TEST
            b_fact[row] -= factor * b[column]
        # great, we've fixed that value. Now let's move on to the next one'
    # now we should be looking at our diagonal, which needs to be 1
    # normalize it if not
    current_value = M_fact[row, row]
    if current_value != 1:
        for column in range(columns):
            M_fact[row, column] /= current_value
```
