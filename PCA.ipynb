PCA Principle Component Analysis

```python
import numpy as np
```

start with a matrix of data

```python
x1 = [4, 8, 13, 7]
x2 = [11, 4, 5, 14]
```

normalize the data centered at 0
this involves calculating and subtracting the means
and then dividing by standard deviation

```python
x1_m = np.mean(x1)
x2_m = np.mean(x2)

x1_sd = np.std(x1, ddof=1)
x2_sd = np.std(x2, ddof=1)
```

```python
x1_norm = ( x1 - x1_m ) / x1_sd
x2_norm = ( x2 - x2_m ) / x2_sd
```

```python
x1_norm
```

get the cooresponding covariance matrix

```python
x1_x2_norm_cov = np.sum(x1_norm*x2_norm) / (len(x1_norm) - 1)
```

```python
x1_norm_var = 1
x2_norm_var = 1
x1_norm_mean = 0
x2_norm_mean = 0
```

```python
cov_mat = [[x1_norm_var, x1_x2_norm_cov],
           [x1_x2_norm_cov, x2_norm_var]]
```

```python
cov_mat
```

```python
cov_mat2 = np.cov(x1_norm, x2_norm)
```

```python
cov_mat2
```

find the eigenvalues

```python
eigenvalues, eigenvectors = np.linalg.eig(cov_mat2)
```

choose the largest eigenvalue and find the eigenvector.
normalize it to distance 1
that's PC1

```python
max_index = np.argmax(eigenvalues)
PC1 = eigenvectors[:,max_index]
```

```python
PC1
```

multiply PC1 by your normalized data to get the new distribution

Calculate Scores or projections of the normalized data onto PC1

X^T * V

```python
PC1_scores = x1_norm * PC1[0] + x2_norm * PC1[1]
```

```python
PC1_scores
```

```python
X = np.column_stack([x1_norm, x2_norm])
PC1_scores = X @ PC1
```
