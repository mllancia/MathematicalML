```python
x1 = np.array([5,4,6,6,7])
x2 = np.array([1,1,0,0,0])
```

```python
def prob(beta, x):
    return 1 / (1 + np.exp(-(beta[0] + beta[1] * x)))
```

We're gonna need to do numerial method such as gradient descent in order to optimize beta in the above formula

Make a gues for beta. I notice that it seems that x2 lowers as x1 raises, so negative B1, and x1 starts at values of 1, so start with B_0 at 1

```python
beta = np.array([1,-1])
maxiter = 1000
acceptance = 0.1
learning_rate = 0.1
```

the gradients of B are based on a summation of all the original data given our equation we want to optize, it's derivative in terms of both betas, and the current best gueses for beta

Let's iterate many times through the gradient descent process, each time making small updates to betas

```python
for j in range(maxiter):
    dB_0 = 0
    dB_1 = 0
    for i in range(len(x1)):
        x = x1[i]
        y = x2[i]
        p = prob(beta, x)

        eq = y * np.log(p) + (1 - y) * np.log(1 - p)

        dB_0 = dB_0 + (y - p)
        dB_1 = dB_1 + (y - p) * x

    if np.sqrt(dB_0**2 + dB_1**2) < acceptance:
        converged = True
        print("converged at iter " + str(j))
        break
    else:
        beta = beta + learning_rate * np.array([dB_0, dB_1])
```
